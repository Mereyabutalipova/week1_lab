### Problem 1: Working with Data [cite: 10]

### Problem 3: Rule-Based Decision [cite: 25]

```python
def pass_fail(score):
    return "Pass" if score >= 60 else "Fail" # [cite: 27, 28]

test_scores = [45, 67, 82, 59] # [cite: 30]
for s in test_scores:
    print(f"Score {s}: {pass_fail(s)}")
```

### Problem 4: Observing Patterns [cite: 31]

```python
sizes = [500, 800, 1000, 1200] # [cite: 33, 34]
prices = [50, 80, 100, 130] # [cite: 33, 34]
for s, p in zip(sizes, prices):
    print(f"Size: {s}, Price: {p}") # [cite: 35]
if all(prices[i] < prices[i+1] for i in range(len(prices)-1)):
    print("Positive trend") # [cite: 37]
```

### Problem 6: Data Categorization [cite: 47]

```python
scores = [35, 58, 62, 74, 89, 45] # [cite: 49]
for s in scores:
    if s < 50: cat = "Low"
    elif 50 <= s <= 70: cat = "Medium"
    else: cat = "High"
    print(f"Score: {s}, Category: {cat}") # [cite: 52]
```

### Problem 7: Simple Dataset Statistics [cite: 54]

```python
ages = [18, 20, 22, 24, 26] # [cite: 57, 58]
heights = [160, 165, 170, 175, 180] # [cite: 59, 60]
for a, h in zip(ages, heights): print(f"Age: {a}, Height: {h}") # [cite: 61]
print("Avg Age:", sum(ages)/len(ages)) # [cite: 62]
print("Avg Height:", sum(heights)/len(heights)) # [cite: 63]
print("Comment: Height increases with age.") # [cite: 64]
```
